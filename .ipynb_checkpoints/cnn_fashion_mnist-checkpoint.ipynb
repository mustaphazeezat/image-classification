# Convolutional neural network CNN for Image Classification

```python
import tensorflow as tf
from tensorflow import keras
from keras import layers
from keras.optimizers import SGD
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

keras.utils.set_random_seed(42)
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.fashion_mnist.load_data()
```

```python
print(np.unique(y_train))
```

**What do these numbers mean?**

According to the [Fashion MNIST Github site](https://github.com/zalandoresearch/fashion-mnist#labels), this is what each number 0-9 corresponds to:

| Label  | Description  |
|---|---|
|  0 | T-shirt/top  |
|  1 |  Trouser |
|  2 |   Pullover |
|  3 |  Dress |
|  4 |  Coat |
|  5 |  Sandal |
|  6 |   Shirt |
|  7 |  Sneaker |
|  8 |   Bag |
|  9 |  Ankle boot |

```python
labels = ["T-shirt/top",
          "Trouser",
          "Pullover",
          "Dress",
          "Coat",
          "Sandal",
          "Shirt",
          "Sneaker",
          "Bag",
          "Ankle boot"]
```

```python
## Helper functions
def plot_loss_curves(history):
  plt.clf()
  history_dict = history.history
  loss_values = history_dict["loss"]
  val_loss_values = history_dict["val_loss"]
  epochs = range(1, len(loss_values) + 1)
  plt.plot(epochs, loss_values, "bo", label="Training loss")
  plt.plot(epochs, val_loss_values, "b", label="Validation loss")
  plt.title("Training and validation loss")
  plt.xlabel("Epochs")
  plt.ylabel("Loss")
  plt.legend()
  plt.show()


def plot_acc_curves(history):
  plt.clf()
  history_dict = history.history
  acc = history_dict["accuracy"]
  val_acc = history_dict["val_accuracy"]
  epochs = range(1, len(acc) + 1)
  plt.plot(epochs, acc, "bo", label="Training acc")
  plt.plot(epochs, val_acc, "b", label="Validation acc")
  plt.title("Training and validation accuracy")
  plt.xlabel("Epochs")
  plt.ylabel("Accuracy")
  plt.legend()
  plt.show()
```

### Data preprocessing

The data ranges from 0-255 because the images are black and white  
to ensure the numbers are not too far from each other 
- Prevents Feature Dominance
- To ensure faster convergence

Normalize to the 0-1 range by dividing everything by 255.

```python
x_train = x_train.astype("float32") / 255
x_test = x_test.astype("float32") / 255
```

```python
x_train.shape
```

CNN requires knowing the type of images we are working with,  
- If the images are black and white we need to add an additional dimension of 1 representing(0-255) of black 
- If the images are colored we add 3 as the dimension representing red, blue and green.

We are working with monocrome images, so we extend it to have 1 dimension.

```python
x_train = np.expand_dims(x_train, -1)
x_test = np.expand_dims(x_test, -1)
```

```python
x_train.shape
```

## The CNN model

```python
## Model funtion

def create_CNN_model(x_train):
    input = keras.Input(shape=x_train.shape[1:])

    # the first convolutional block
    x = keras.layers.Conv2D(64,                    # Number of filters
                        kernel_size=(3, 3),    # The shape of each filter
                        activation="relu",     # RELU activation as usual
                        name="Conv_1")(input)
    # pooling layer
    x = keras.layers.MaxPool2D()(x)
    # end of first convolutional block
    
    # the second convolutional block
    # convolutional layer
    x = keras.layers.Conv2D(32,                    # Number of filters
                            kernel_size=(2, 2),    # The shape of each filter
                            activation="relu",     # RELU activation as usual
                            name="Conv_2")(x)
    # pooling layer
    x = keras.layers.MaxPool2D()(x)
    # end of second convolutional block


    # flatten layer
    x = keras.layers.Flatten()(x)

    # fully-connected (dense) ReLU layer
    x = keras.layers.Dense(256, activation="relu")(x)

    x = keras.layers.Dropout(0.2)(x)

    # output softmax layer
    output = keras.layers.Dense(10, activation="softmax")(x)
    
    
    
    model = keras.Model(input, output)
    

    model.compile(loss='sparse_categorical_crossentropy',
             optimizer= "adam",
             metrics=['accuracy'])
    
    model.summary()
    
    return model
```

```python
model = create_CNN_model(x_train)
```

```python
history = model.fit(x_train,
                    y_train,
                    batch_size=64,
                    epochs=10,
                    validation_split=0.2)
```

## Model evaluation

```python
plot_loss_curves(history)
```

Based on this plot the training loss reduced as epoch reduces,   
indicating that the model improves based on exploring the same dataset over and over again.

The validation loss was also reducing as the training loss but stop improving around the 6th epoch  
and slightly begin to rise around the 9th epoch which is an indication of beginning of overfitting.

```python
plot_acc_curves(history)
```

Based on this accuracy plot,  
The **training accuracy** improved based on exploring the same dataset over and over again.  

The **validation accuracy** kept increasing with the epoch but at the 3rd epoch,  
the validation accuracy improve slowly but at 9 epoch the validation accuraccy increases again  

The model Accuraccy increasing at 9 epoch, **cancels the notion of overfitting**



Evaluating the model based on Training data does not state how afficient the model is.  

Evaluation on unseen data in the model, shows how well the model generalise on unseen data.   

We use the test set for the model evaluation

```python
score = model.evaluate(x_test, y_test)
print("Test accuracy:", score[1])
```

##### This CNN model have 90% accuracy on generalisation on unseen data.

```python
# get the predictions
y_hat = np.argmax(model.predict(x_test), axis=1)

# collect examples where the model made a mistake
misses = np.where(y_hat != y_test)
```

```python
# Plot a random sample of 25 test images incorrectly classified by the model,
# their predicted labels and ground truth
figure = plt.figure(figsize=(20, 8))
for i, index in enumerate(np.random.choice(misses[0], size=25, replace=False)):
    ax = figure.add_subplot(5, 5, i + 1, xticks=[], yticks=[])
    # Display each image
    ax.imshow(np.squeeze(x_test[index]), cmap="gray")
    # Set the title for each image
    ax.set_title("{} ({})".format(labels[y_hat[index]],
                                  labels[y_test[index]]))
```

```python
actuals = [labels[i] for i in y_test]
predictions = [labels[i] for i in y_hat]

df = pd.DataFrame({'Predictions': predictions, 'Actuals': actuals})
pd.crosstab(df.Predictions, df.Actuals)
```

```python
## Model funtion

def create_CNN_model_aug(x_train):
    input = keras.Input(shape=x_train.shape[1:])

    x = keras.Sequential([
        keras.layers.RandomTranslation(0.1, 0.1),
        keras.layers.RandomRotation(0.05),
        keras.layers.RandomZoom(
            height_factor=(-0.1, 0.1),
            width_factor=(-0.1, 0.1)
        )
        
    ], name="data_augmentation")(input)

    # the first convolutional block

    x = keras.layers.Conv2D(32,                    # Number of filters
                            kernel_size=(3, 3),    # The shape of each filter
                            padding="same",
                            activation="relu",     # RELU activation as usual
                            name="Conv_1")(input)
    # pooling layer
    x = keras.layers.MaxPool2D()(x)
   
    # end of first convolutional block
    
    # the second convolutional block
    # convolutional layer
    x = keras.layers.Conv2D(64,                    # Number of filters
                        kernel_size=(3, 3),    # The shape of each filter
                        padding="same",
                        activation="relu",     # RELU activation as usual
                        name="Conv_2")(x)
    # pooling layer
    x = keras.layers.MaxPool2D()(x)
    # end of second convolutional block


    # flatten layer
    x = keras.layers.Flatten()(x)

    # fully-connected (dense) ReLU layer
    x = keras.layers.Dense(256, activation="relu")(x)

    x = keras.layers.Dropout(0.2)(x)

    # output softmax layer
    output = keras.layers.Dense(10, activation="softmax")(x)
    
    
    
    model = keras.Model(input, output)
    

    model.compile(loss='sparse_categorical_crossentropy',
             optimizer= "adam",
             metrics=['accuracy'])
    
    return model
```

```python
model_aug = create_CNN_model_aug(x_train)
```

```python
early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)

history = model_aug.fit(
    x_train,
    y_train,
    batch_size=64,
    epochs=30,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
score = model_aug.evaluate(x_test, y_test)
print("Test accuracy:", score[1])
```
